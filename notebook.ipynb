## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Berapa persentase pertumbuhan atau penurunan jumlah pesanan secara bulanan (Month-over-Month) pada platform e-commerce Olist selama periode Januari 2017 hingga Agustus 2018, dan pada bulan apa saja terjadi penurunan lebih dari 10%?
- **Pertanyaan 2:** Dari seluruh kategori produk yang dijual di platform Olist selama 2017-2018, kategori mana saja yang memiliki jumlah pesanan di atas 1.000 namun rata-rata skor review di bawah 4.0?
- **Pertanyaan 3:** Berapa persentase pelanggan Olist yang termasuk dalam segmen 'At Risk' dan 'Lost' berdasarkan analisis RFM (Recency, Frequency, Monetary) selama periode 2017-2018?
- **Pertanyaan 4:** Negara bagian (state) mana di Brasil yang memiliki jumlah pelanggan di atas 2.000 namun rata-rata spending per pelanggan di bawah rata-rata nasional selama 2017-2018?

## Exploratory Data Analysis (EDA)

### Explore Jumlah pesanan berdasarkan Status Pesanan

```python
# 1. Jumlah pesanan berdasarkan Status Pesanan
plt.figure(figsize=(10, 6))
status_counts = orders_df['order_status'].value_counts().head(5)

sns.barplot(x=status_counts.values, y=status_counts.index, palette='Blues_r')
plt.title('Jumlah pesanan berdasarkan Status Pesanan', fontsize=16)
plt.xlabel('Jumlah Pesanan', fontsize=12)
plt.ylabel('Status', fontsize=12)

# Menambahkan angka di sebelah bar
for i, v in enumerate(status_counts.values):
    plt.text(v + 500, i, f"{v:,.0f}", color='black', va='center')

plt.show()
```

```python
# 1. Total Pesanan & Pesanan Berhasil
total_orders = orders_df['order_id'].nunique()
delivered_orders = orders_df[orders_df['order_status'] == 'delivered']['order_id'].nunique()
delivery_rate = (delivered_orders / total_orders) * 100
```

```python
print(f"Total Pesanan Keseluruhan: {total_orders:,.0f}")
print(f"Total Pesanan Berhasil (Delivered): {delivered_orders:,.0f} ({delivery_rate:.2f}%)")
```

### Explore Total Pendapatan berdasarkan Metode Pembayaran

```python
# 2. Total Pendapatan berdasarkan Metode Pembayaran
plt.figure(figsize=(10, 6))
payment_revenue = order_payments_df.groupby('payment_type')['payment_value'].sum().sort_values(ascending=False)

sns.barplot(x=payment_revenue.values, y=payment_revenue.index, palette='Greens_r')
plt.title('Total Pendapatan berdasarkan Metode Pembayaran', fontsize=16)
plt.xlabel('Total Nilai (R$)', fontsize=12)
plt.ylabel('Metode Pembayaran', fontsize=12)

# Menambahkan angka di sebelah bar
for i, v in enumerate(payment_revenue.values):
    plt.text(v + 100000, i, f"R$ {v:,.0f}", color='black', va='center')

plt.show()
```

```python
# 2. Total Pendapatan (Gross Merchandise Value / GMV)
total_revenue = order_payments_df['payment_value'].sum()
print(f"Total Pendapatan (GMV): R$ {total_revenue:,.2f}")
```

**Insight:**
- Grafik ini menunjukkan fakta yang sangat mencolok bahwa mayoritas mutlak perputaran uang (GMV) di platform Olist berasal dari transaksi kartu kredit (credit_card), yang menembus angka raksasa R$ 12.542.084. Angka ini membuktikan bahwa kartu kredit adalah urat nadi utama pendapatan perusahaan.

```python
# 1. Menggabungkan data Pelanggan, Pesanan, dan Pembayaran
# Filter pesanan yang sudah sukses (delivered)
delivered_orders = orders_df[orders_df['order_status'] == 'delivered'].copy()

# Gabung orders dengan customers (untuk mendapatkan ID asli/customer_unique_id)
rfm_base = pd.merge(delivered_orders[['order_id', 'customer_id', 'order_purchase_timestamp']],
                    customers_df[['customer_id', 'customer_unique_id']],
                    on='customer_id', how='inner')

# 3. Gabung dengan payments (untuk menghitung Monetary)
# totalkan terlebih dahulu di payment_value per pesanan (karena 1 pesanan bisa 2x bayar misal pakai voucher + kartu)
order_payments_agg = order_payments_df.groupby('order_id')['payment_value'].sum().reset_index()

# Gabungkan ke tabel dasar
rfm_base = pd.merge(rfm_base, order_payments_agg, on='order_id', how='inner')

print("-> Tabel dasar RFM berhasil dibuat (menghubungkan identitas, waktu beli, dan jumlah uang).")
display(rfm_base[['customer_unique_id', 'order_purchase_timestamp', 'payment_value']].head(5))
print("-" * 50)
```

```python
# 2. Menghitung nilai Recency, Frequency, dan Monetary (RFM)
# Pastikan tipe data tanggal sudah benar
rfm_base['order_purchase_timestamp'] = pd.to_datetime(rfm_base['order_purchase_timestamp'])

# Tentukan tanggal 'Hari Ini' (Snapshot Date) yaitu H+1 dari tanggal transaksi terakhir di database
snapshot_date = rfm_base['order_purchase_timestamp'].max() + dt.timedelta(days=1)

# Agregasi data menjadi 1 baris per pelanggan (customer_unique_id)
rfm_df = rfm_base.groupby('customer_unique_id').agg({
    'order_purchase_timestamp': lambda x: (snapshot_date - x.max()).days, # RECENCY (Selisih hari)
    'order_id': 'nunique',                                                # FREQUENCY (Banyak transaksi)
    'payment_value': 'sum'                                                # MONETARY (Total uang)
}).reset_index()

# Ubah nama kolom agar sesuai
rfm_df.rename(columns={
    'order_purchase_timestamp': 'Recency',
    'order_id': 'Frequency',
    'payment_value': 'Monetary'
}, inplace=True)

print("-> Nilai mentah R, F, M untuk setiap pelanggan berhasil dihitung.")
display(rfm_df.head(5))
print("-" * 50)
```

```python
# 3. Memberikan Skor (Scoring) 1 sampai 5
# Semakin tinggi skor, semakin bagus.

# 1. Recency Score: Dibagi 5 kelompok. Hari paling kecil (baru beli) dapat skor 5.
rfm_df['R_Score'] = pd.qcut(rfm_df['Recency'], 5, labels=[5, 4, 3, 2, 1])

# 2. Monetary Score: Dibagi 5 kelompok. Belanja paling banyak dapat skor 5.
rfm_df['M_Score'] = pd.qcut(rfm_df['Monetary'], 5, labels=[1, 2, 3, 4, 5])

# 3. Frequency Score: dibuat manual karena mayoritas belanja 1x.
# Belanja 1x = skor 1. Belanja 2x = skor 2. Belanja 3x atau lebih = skor 3.
def calc_f_score(x):
    if x == 1: return 1
    elif x == 2: return 2
    else: return 3
rfm_df['F_Score'] = rfm_df['Frequency'].apply(calc_f_score)

print("-> Skor R, F, M berhasil dikonversi menjadi skala peringkat.")
display(rfm_df[['customer_unique_id', 'R_Score', 'F_Score', 'M_Score']].head(5))
print("-" * 50)
```

```python
# 4. Segmentasi Pelanggan ('At Risk' & 'Lost')
# Dalam RFM, retensi paling dipengaruhi oleh "Recency" (R).
# Kita buat aturan sederhana berdasarkan R_Score:
# Skor 1 = 'Lost' (Sudah sangat lama tidak kembali)
# Skor 2 = 'At Risk' (Hampir hilang)
# Skor 3, 4, 5 = 'Active' (Masih berbelanja baru-baru ini)

def assign_segment(r_score):
    if r_score == 1:
        return 'Lost'
    elif r_score == 2:
        return 'At Risk'
    else:
        return 'Active'

rfm_df['Segment'] = rfm_df['R_Score'].apply(assign_segment)

# Menghitung persentase setiap segmen
segment_counts = rfm_df['Segment'].value_counts().reset_index()
segment_counts.columns = ['Segment', 'Total_Customers']
segment_counts['Percentage (%)'] = (segment_counts['Total_Customers'] / len(rfm_df)) * 100

print("-> Segmentasi berhasil dilakukan. Berikut adalah distribusi populasinya:")
display(segment_counts.round(2))
print("=" * 50)
```

- **Conclusion pertanyaan 1:** Sepanjang 2017-2018, tren pesanan menunjukkan pertumbuhan positif dengan puncak tertinggi pada November 2017 (+62.8%). Namun, bisnis sangat rentan terhadap kelesuan di pertengahan tahun, dibuktikan dengan penurunan pesanan di atas 10% pada bulan April 2018 dan Desember 2018.
- **Conclusion pertanyaan 2:** Terdapat anomali performa pada kategori utama penyumbang pesanan, yaitu* bed_bath_table, furniture_decor, dan computers_accessories*. Meskipun mencetak ribuan transaksi, ketiga kategori ini  rata-rata skor ulasan di bawah standar (skor < 4.0), yang mengindikasikan tingginya masalah kualitas barang atau kerusakan saat pengiriman logistik.
- **Conclusion pertanyaan 3:** Platform Olist didominasi oleh pembeli satu kali (one-time buyer). Analisis RFM mengonfirmasi tingginya tingkat churn, di mana nyaris 40% basis pelanggan saat ini berstatus 'Lost' (20.0%) dan 'At Risk' (19.9%).
- **Conclusion pertanyaan 4:** : Terdapat ketimpangan daya beli yang krusial di dua negara bagian utama, yaitu São Paulo dan Paraná. Meskipun kedua wilayah ini memiliki populasi pelanggan yang masif (> 2.000 pengguna), rata-rata pengeluaran mereka (terutama São Paulo di angka R 147) masih tertinggal di bawah rata-rata pengeluaran nasional sebesar R$ 165.

**Rekomendasi Action Item:**
- **Inisiasi Kampanye Pertengahan Tahun (Mid-Year Sale):** Tim Marketing tidak boleh hanya bergantung pada momentum kuartal keempat (Black Friday). Buatlah kampanye khusus di kuartal kedua (April - Juni) untuk menjaga stabilitas volume pesanan bulanan dan mencegah tren penurunan yang tajam.

- **Audit Operasional & Pengetatan Standar Kemasan:** Tim Operasional (Seller Center) wajib memanggil dan mengevaluasi penjual di kategori bed_bath_table dan furniture_decor. Terapkan standar Quality Control (QC) dan wajib kemasan ekstra aman (extra bubble wrap / kayu) untuk barang pecah belah atau berdimensi besar demi memperbaiki kepuasan ulasan pelanggan.

- **Eksekusi Kampanye Penyelamatan Pelanggan (Retention Program):** Jangan membuang budget hanya untuk mencari pelanggan baru. Segera luncurkan kampanye promosi terarah (targeted promo) seperti pengiriman e-mail "Voucher Diskon Welcome Back" yang dikhususkan secara eksklusif bagi 19.9% pelanggan di segmen 'At Risk' agar mereka kembali melakukan transaksi.

- **Optimalisasi Nilai Keranjang Belanja (Upselling Wilayah Target):** Eksekusi taktik upselling khusus untuk wilayah São Paulo dan Paraná. Berikan subsidi "Gratis Ongkir untuk Minimal Belanja R$ 175" khusus pengiriman ke dua wilayah tersebut. Strategi ini dirancang untuk memaksa pelanggan menambah item di keranjang mereka, sehingga mendongkrak rata-rata daya beli mereka (khususnya pasar raksasa São Paulo) agar melewati angka rata-rata nasional.
